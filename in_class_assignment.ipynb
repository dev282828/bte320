```python
def reverse(s):
    if len(s) == 0:
        return s
    else:
        return reverse(s[1:]) + s[0]

print(reverse('hello'))
print(reverse('12345'))
```

```python
def loan_payment(principal, annual_rate, years):
    assert (isinstance(principal, (int, float)) and principal >= 0), "Principal must be a non-negative number."
    assert (isinstance(annual_rate, (int, float)) and annual_rate >= 0), "Annual rate must be a non-negative number."
    assert (isinstance(years, (int, float)) and years > 0), "Years must be a positive number."

    try:
        r = (annual_rate / 100) / 12
        n = years * 12

        if r > 0:
            payment = (r * principal) / (1 - (1 + r)**(-n))
        else:
            payment = principal / n

        return payment

    except ZeroDivisionError:
        raise ValueError("Cannot calculate payment. Check if years/rate are valid.")
    except Exception as e:
        raise ValueError(f"An unexpected error occurred: {e}")

p = float(input("Enter loan amount: "))
rate = float(input("Enter annual interest rate (%): "))
y = float(input("Enter number of years: "))

monthly_pay = loan_payment(p, rate, y)
print(f"Monthly Payment: ${monthly_pay:.2f}")
```

```python
def getValues(n):
    items = []
    for i in range(n):
        items.append(float(input(f"Value {i+1}: ")))
    return items

def calcAverage(alist):
    try:
        subset = alist[3:]
        return sum(subset) / len(subset)
    except ZeroDivisionError:
        return "Not enough items"

def main():
    n = int(input("Enter number of items: "))
    if n > 3:
        alist = getValues(n)
        print(f"Average: {calcAverage(alist)}")
    else:
        print("Need more than 3 items")
main()
```

```python
class Greeting:
    def __init__(self, first, last):
        self.first = first
        self.last = last

    def display(self, formal=False):
        if formal:
            print(f"{self.last}, {self.first}")
        else:
            print(self.first)

    def __str__(self):
        return "Instance of class 'Greeting'"

name = Greeting("Jane", "Doe")
print(name)
name.display()
name.display(formal=True)
```

```python
class Student:
    def __init__(self, name, number, courses=None):
        self.__name = name
        self.number = number

        self.__courses = courses if courses is not None else []

    def get_name(self):
        return self.__name

    def get_courses(self):
        return self.__courses

    def set_name(self, newName):
        assert isinstance(newName, str), "Name must be a string"
        self.__name = newName

    def __len__(self):
        return len(self.__courses)

    def enrol(self, newCourse):
        if newCourse not in self.__courses:
            self.__courses.append(newCourse)
        else:
            print(f"{self.__name} is already enrolled in {newCourse}")


s = Student("Jane Doe", 12345)
print(f"Student name: {s.get_name()}")
s.enrol("Math")
s.enrol("Math")
print(f"Number of courses: {len(s)}")




```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self, other):
        return ((self.x - other.x)**2 + (self.y - other.y)**2)**0.5

class Rectangle:
    def __init__(self, p1, p2, p3, p4):

        self.p1 = p1
        self.p2 = p2
        self.p3 = p3
        self.p4 = p4


        self.width = self.calc_width()
        self.height = self.calc_height()

    def calc_width(self):
        return self.p1.distance(self.p2)

    def calc_height(self):
        return self.p2.distance(self.p3)

    def area(self):
        return self.width * self.height

pt1 = Point(0, 0)
pt2 = Point(10, 0)
pt3 = Point(10, 5)
pt4 = Point(0, 5)

rect = Rectangle(pt1, pt2, pt3, pt4)

print(f"Width: {rect.width}")
print(f"Height: {rect.height}")
print(f"Area: {rect.area()}")
```

```python
class Employee:
    def __init__(self, name, wage, hours):
        self.name = name
        self.__wage = wage
        self.hours = hours

    def get_wage(self):
        return self.__wage

    def set_wage(self, new_wage):
        assert isinstance(new_wage, (int, float))
        self.__wage = new_wage

    def calculate_pay(self):
        return self.__wage * self.hours

    def display_info(self):
        print(f"Name: {self.name}")
        print(f"Wage: {self.__wage}")
        print(f"Hours: {self.hours}")
        print(f"Pay: {self.calculate_pay()}")

class Manager(Employee):
    def __init__(self, name, wage, hours, bonus):
        super().__init__(name, wage, hours)
        self.bonus = bonus

    def calculate_pay(self):

        return super().calculate_pay() + self.bonus

employee = Employee('Bob', 50, 10)
manager = Manager('Alice', 100, 15, 500)

print(f"Employee {employee.name} pay: {employee.calculate_pay()}$")
print(f"Manager {manager.name} pay: {manager.calculate_pay()}$")
```
